```python
import json
import os
import pandas as pd
import subprocess

result = subprocess.run(["python", "../src/data_processing.py"], capture_output=True, text=True)
print(result.stdout)
print(result.stderr)

# ─── 1. Load a single match file to inspect structure ───────────────────────
DATA_FOLDER = "/Users/aryanjungchhetri/Developers/6th sem/individual/data/raw/Syed_Mushtaq_Ali_Trophy_Raw"  # <-- change this

# Pick the first file to explore
sample_file = [f for f in os.listdir(DATA_FOLDER) if f.endswith(".json")][0]

with open(os.path.join(DATA_FOLDER, sample_file), "r") as f:
    match = json.load(f)

# ─── 2. Explore top-level keys ───────────────────────────────────────────────
print("Top-level keys:", list(match.keys()))
# Expected: ['meta', 'info', 'innings']

# ─── 3. Match metadata ───────────────────────────────────────────────────────
info = match["info"]
print("\n--- Match Info ---")
print("Teams     :", info["teams"])
print("Date      :", info["dates"])
print("Venue     :", info.get("venue", "N/A"))
print("City      :", info.get("city", "N/A"))
print("Toss      :", info["toss"])
print("Outcome   :", info.get("outcome", "N/A"))

# ─── 4. Explore innings & deliveries ────────────────────────────────────────
print("\n--- Innings Structure ---")
for i, inning in enumerate(match["innings"]):
    print(f"\nInnings {i+1}: {inning['team']}")
    first_over = inning["overs"][0]
    print("  First over deliveries sample:")
    for delivery in first_over["deliveries"][:2]:
        print("  ", delivery)


# ─── 5. Check for extras, wickets, and player registry ──────────────────────

print("=== INFO KEYS ===")
print(list(info.keys()))

print("\n=== PLAYERS REGISTRY (if exists) ===")
print(info.get("registry", "No registry found"))
print(info.get("players", "No players key found"))

print("\n=== SAMPLE DELIVERY WITH WICKET ===")
# Find a delivery that has a wicket in it
found = False
for inning in match["innings"]:
    for over in inning["overs"]:
        for delivery in over["deliveries"]:
            if "wickets" in delivery:
                print(delivery)
                found = True
                break
        if found:
            break
    if found:
        break

print("\n=== SAMPLE DELIVERY WITH EXTRAS ===")
found = False
for inning in match["innings"]:
    for over in inning["overs"]:
        for delivery in over["deliveries"]:
            if delivery["runs"]["extras"] > 0:
                print(delivery)
                found = True
                break
        if found:
            break
    if found:
        break

print("\n=== TOTAL FILES IN FOLDER ===")
all_files = [f for f in os.listdir(DATA_FOLDER) if f.endswith(".json")]
print(f"{len(all_files)} match files found")



# ─── 6. Inspect registry, players, wicket structure & season info ────────────

print("=== SEASON & EVENT ===")
print("Season:", info.get("season"))
print("Event :", info.get("event"))

print("\n=== PLAYERS PER TEAM ===")
for team, players in info["players"].items():
    print(f"\n{team}:")
    print(players)

print("\n=== REGISTRY SAMPLE (first 5) ===")
registry = info.get("registry", {}).get("people", {})
for i, (name, pid) in enumerate(registry.items()):
    print(f"  {name}: {pid}")
    if i == 4:
        break

print("\n=== WICKET DELIVERY STRUCTURE ===")
found = False
for inning in match["innings"]:
    for over in inning["overs"]:
        for delivery in over["deliveries"]:
            if "wickets" in delivery:
                print(delivery)
                found = True
                break
        if found:
            break
    if found:
        break

print("\n=== OVER STRUCTURE (over number) ===")
first_inning = match["innings"][0]
for over in first_inning["overs"][:3]:
    print(f"Over {over['over']} → {len(over['deliveries'])} deliveries")
```
