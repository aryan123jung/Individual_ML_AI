```python
import json
import os
import pandas as pd

pd.set_option("display.max_columns", 30)
pd.set_option("display.width", 200)

IPL_FOLDER = "/Users/aryanjungchhetri/Developers/6th sem/individual/data/raw/ipl_raw"

# ─── Total files ──────────────────────────────────────────────────────────────
all_files = [f for f in os.listdir(IPL_FOLDER) if f.endswith(".json")]
print(f"Total IPL match files: {len(all_files)}")

# ─── Load one sample file ─────────────────────────────────────────────────────
sample_file = all_files[0]
with open(os.path.join(IPL_FOLDER, sample_file), "r") as f:
    match = json.load(f)

# ─── Top level keys ───────────────────────────────────────────────────────────
print("\nTop-level keys:", list(match.keys()))

# ─── Match info ───────────────────────────────────────────────────────────────
info = match["info"]
print("\n=== INFO KEYS ===")
print(list(info.keys()))

print("\n--- Match Info ---")
print("Teams   :", info["teams"])
print("Date    :", info["dates"])
print("Season  :", info.get("season"))
print("Venue   :", info.get("venue"))
print("Toss    :", info["toss"])
print("Outcome :", info.get("outcome"))
print("Event   :", info.get("event"))

# ─── Innings & deliveries ─────────────────────────────────────────────────────
print("\n=== INNINGS STRUCTURE ===")
for i, inning in enumerate(match["innings"]):
    print(f"\nInnings {i+1}: {inning['team']}")
    first_over = inning["overs"][0]
    print(f"  Over 0 → {len(first_over['deliveries'])} deliveries")
    for delivery in first_over["deliveries"][:2]:
        print("  ", delivery)

# ─── Wicket sample ────────────────────────────────────────────────────────────
print("\n=== SAMPLE WICKET DELIVERY ===")
found = False
for inning in match["innings"]:
    for over in inning["overs"]:
        for delivery in over["deliveries"]:
            if "wickets" in delivery:
                print(delivery)
                found = True
                break
        if found:
            break
    if found:
        break

# ─── Extras sample ────────────────────────────────────────────────────────────
print("\n=== SAMPLE EXTRAS DELIVERY ===")
found = False
for inning in match["innings"]:
    for over in inning["overs"]:
        for delivery in over["deliveries"]:
            if delivery["runs"]["extras"] > 0:
                print(delivery)
                found = True
                break
        if found:
            break
    if found:
        break
```
