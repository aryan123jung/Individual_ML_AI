```python
import pandas as pd
import os

# ─── CONFIG ──────────────────────────────────────────────────────────────────
pd.set_option("display.max_rows", 100)
pd.set_option("display.max_columns", 30)
pd.set_option("display.width", 200)

BASE = "/Users/aryanjungchhetri/Developers/6th sem/individual"

# ─── LOAD ─────────────────────────────────────────────────────────────────────
batting_df = pd.read_csv(f"{BASE}/data/processed/smat_batting_raw.csv")
bowling_df = pd.read_csv(f"{BASE}/data/processed/smat_bowling_raw.csv")

print("✅ Reloaded")
print(f"Batting : {batting_df.shape}")
print(f"Bowling : {bowling_df.shape}")

# ─── NULLS ────────────────────────────────────────────────────────────────────
print("\n=== BATTING NULLS ===")
print(batting_df.isnull().sum().to_string())

print("\n=== BOWLING NULLS ===")
print(bowling_df.isnull().sum().to_string())

# ─── SEASONS ──────────────────────────────────────────────────────────────────
print("\n=== SEASONS (matches per season) ===")
print(batting_df.groupby("season")["match_file"].nunique().to_string())

# ─── TOP SCORERS ──────────────────────────────────────────────────────────────
print("\n=== TOP 10 RUN SCORERS (all seasons) ===")
print(batting_df.groupby("player")["runs"].sum().sort_values(ascending=False).head(10).to_string())

# ─── TOP WICKET TAKERS ────────────────────────────────────────────────────────
print("\n=== TOP 10 WICKET TAKERS (all seasons) ===")
print(bowling_df.groupby("player")["wickets"].sum().sort_values(ascending=False).head(10).to_string())

# ─── ZERO BALLS FACED ─────────────────────────────────────────────────────────
print("\n=== BATTERS WITH 0 BALLS FACED ===")
zero_balls = batting_df[batting_df["balls_faced"] == 0]
print(f"Count : {len(zero_balls)}")
print(zero_balls[["player", "runs", "balls_faced", "dismissed"]].head(10).to_string())

# ─── BASIC STATS SANITY CHECK ─────────────────────────────────────────────────
print("\n=== BATTING BASIC STATS ===")
print(batting_df[["runs", "balls_faced", "fours", "sixes", "dismissed"]].describe().to_string())

print("\n=== BOWLING BASIC STATS ===")
print(bowling_df[["runs_conceded", "balls_bowled", "wickets", "wides", "noballs"]].describe().to_string())

# ─── UNIQUE PLAYERS ───────────────────────────────────────────────────────────
print("\n=== UNIQUE PLAYERS ===")
print(f"Unique batters  : {batting_df['player'].nunique()}")
print(f"Unique bowlers  : {bowling_df['player'].nunique()}")
print(f"Unique teams    : {batting_df['team'].nunique()}")
print(f"Unique venues   : {batting_df['venue'].nunique()}")
```
